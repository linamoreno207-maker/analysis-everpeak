```python
df ["skip_rate"].head() #La estructura inicial "df[nombre_columna]"me trae una columna especifica y .head lo que hace es traerme un numero de filas, cuando es por defecto () trae solo 5
```

```python
df["skip_porcentaje"] = df["skip_rate"]*100
df["skip_p_2"] =(df["skip_rate"]) * 200 #esta es una columna más que se añadió "por error" con el fin de aprender con que funcion se borran columnas, la cual se encuentra en la siguiente sección

df.head()
```

```python
df =df.drop(columns = ["skip_p_2"]) #como se puede observar, ya se elimino la columna skip_p_2
df.head()
```

```python
col = ["artist","release_year","streams_millions"]
df[col]
df1 = df[["artist","release_year","streams_millions"]].copy()#con esta opcion lo que hago es sacar solo algunas columnas que me interesan
df1.head()
```

```python
df1[df1["release_year"]>2023] #data_frame [escribir la condición ]->condicion-> seleccionar una columna aplicar el filtro
```

```python
df1[(df1["release_year"] > 2022) & (df1["streams_millions"] > 300)] #esta es la estructura para crear un FILTRO
```

```python
df.loc[(df["release_year"] > 2022) & (df["streams_millions"] > 300), ["platform"]]  #data_frame [(filas) , (columnas)]
```

```python
# el equipo editorial quiere revisar canciones de spotyfy con mas de 500 millones de reproducciónes y una tasa de abandono inferior al 25%, necesita ver el titulo, artista, país, reproduccones y tasa de abandono
df.loc[(df["platform"] == "Spotify") & (df["skip_porcentaje"] < 25) & (df["streams_millions"] > 500), ["song_title", "artist", "country", "streams_millions", "skip_porcentaje"]]
```
